**Note:** Although the subtomogram size in `./brainweb_config.yaml` is set to $64$, the sub-tomograms ${\color{red}\mathbf{v}^0_{\color{white}0}}$ and ${\color{blue}\mathbf{v}^1_{\color{white}0}}$ are actually of size $92$ (DeepDeWedge-2D rounds $\sqrt{2} \times 64 \approx 90.5$ up to the nearest even integer). As in the original tutorial, this is because for model fitting, the sub-tomograms are randomly rotated — but here the rotation is a 2D rotation of a square patch, not a 3D rotation of a cube. If sub-tomograms of size $64$ were extracted directly, the rotated sub-tomograms would have black borders for some rotations. To avoid this, sub-tomograms are automatically extracted with this larger shape, and then cropped to shape $64$ after rotation during the generation of the model inputs and targets (see next notebook).

## Most Important Parameters of `prepare_data` 

You can find a full list of parameters for `ddw prepare-data` by executing
```
ddw prepare-data --help
```
in the commandline. Here, we only discuss the most important parameters.


The most important parameters are the tomogram pairs (here, the two BrainWeb reconstructions with independent missing wedges) from which the sub-tomogram pairs are extracted, and the size of the sub-tomograms:

````python
print_help_for_function_arguments(prepare_data, print_only_required=True)
````

Another important (but not required) parameter are 2 stride values for the 2D sliding window procedure used to extract the sub-tomograms (one per spatial dimension of the square sub-tomogram). Using smaller strides will result in more sub-tomograms, but also in more overlap between them.

**Note on the BrainWeb strides:** DeepDeWedge-2D first splits each tomogram into 2D slices along one axis (every slice is used, there is no stride along that axis), and only slides a 2D window over the remaining two dimensions. The original 3D tutorial's strides `[16, 19, 16]` therefore become `[16, 16]` here — we keep the strides for the two dimensions that are still tiled with a sliding window, and drop the middle value (`19`), which was the stride along the axis that DeepDeWedge-2D now slices exhaustively instead.

````python
print_help_for_function_arguments(prepare_data, arg_names=["subtomo_extraction_strides"])
````

**Note on masks:** the original tutorial also supports restricting sub-tomogram extraction to a region of interest (ROI) using binary `.mrc` masks (`mask_files` / `min_nonzero_mask_fraction_in_subtomo`). We do **not** use a mask for the BrainWeb data — sub-tomograms are extracted from the full volume.

**Note (DeepDeWedge-2D):** mask-based extraction (`mask_files` / `min_nonzero_mask_fraction_in_subtomo`) is not functional yet in the current 2D implementation — this is fine here since BrainWeb doesn't use masks anyway, but keep it in mind for other datasets.
